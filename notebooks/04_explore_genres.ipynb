# Explore Gooogle Categories and Open Library subjects

There are several tags in the Google Books categories and Open Library subjects that are irrelevant for book recommendation, such as New York Times bestseller tags. I will identify these tags and manually exclude those that do not provide useful information about the book's content or genre.

```python
ENRICHED_FILE = "/home/claraoberle/personal-book-recommender/data/enriched/enriched_books.json"
```

```python
import json

with open(ENRICHED_FILE, "r") as f:
    enriched = json.load(f)

all_genres = set()

for book in enriched.values():
    google_categories = book["google_books"].get("categories") or []

    open_library_subjects = [
        subject["name"]
        for subject in (book["open_library"].get("subjects") or [])
    ]

    all_genres.update(google_categories)
    all_genres.update(open_library_subjects)

print(f"Number of unique genres/subjects: {len(all_genres)}")

for genre in sorted(all_genres):
    print(genre)
```
