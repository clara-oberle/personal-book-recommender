# Sanity check for embeddings.npy

```python
import numpy as np
import pandas as pd
```

```python
BOOK_ID_PATH = "/home/claraoberle/personal-book-recommender/data/embedding/book_ids.npy"
EMBEDDINGD_PATH = "/home/claraoberle/personal-book-recommender/data/embedding/embeddings.npy"
CLEAN_DF_PATH = "/home/claraoberle/personal-book-recommender/data/clean/goodreads_clean.csv"
```

```python
book_ids = np.load(BOOK_ID_PATH, allow_pickle=True)
embeddings = np.load(EMBEDDINGD_PATH, allow_pickle=True)
clean_df = pd.read_csv(CLEAN_DF_PATH)
```

I will check the top similarity books to several read books and see if the match is reasonable.

```python
def print_top_similarities(book_id, embeddings, clean_df, book_ids):
    # find the vector for the hunger games
    idx = np.where(book_ids == book_id)[0][0]
    query_vector = embeddings[idx]

    # Calculate similarity with every other book
    similarities = embeddings @ query_vector

    # make the similarity with itself -1 and sort similarities
    similarities[idx] = -1
    top_indices = np.argsort(similarities)[-5:][::-1]

    results = clean_df[clean_df["Book Id"].astype(str).isin(book_ids[top_indices])].copy()

    results["similarity"] = [
        similarities[np.where(book_ids == str(book_id))[0][0]]
        for book_id in results["Book Id"]
    ]
    
    return results[["Book Id", "Title", "similarity"]].sort_values(
            "similarity", ascending=False
        )
```

```python
the_hunger_games_id = "2767052"
el_señor_de_los_anillos = "222947"
paper_towns = "6442769"
the_selection = "10507293"

print_top_similarities(the_hunger_games_id, embeddings, clean_df, book_ids)
```

```python
print_top_similarities(el_señor_de_los_anillos, embeddings, clean_df, book_ids)
```

```python
print_top_similarities(paper_towns, embeddings, clean_df, book_ids)
```

```python
print_top_similarities(the_selection, embeddings, clean_df, book_ids)
```
